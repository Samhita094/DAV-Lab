```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram
from sklearn.cluster import AgglomerativeClustering


#Load dataset

df = pd.read_csv("Mall_Customers.csv")

print(df.shape)
print(df.head())
print("Missing values:", df.isnull().sum().sum())

#Select two features

X = df[["Annual Income (k$)", "Spending Score (1-100)"]]

#Create dendrogram

Z = linkage(X, method="ward")

plt.figure(figsize=(10, 5))
dendrogram(Z)
plt.title("Dendrogram")
plt.xlabel("Customers")
plt.ylabel("Distance")
plt.show()

#Apply Agglomerative Clustering

model = AgglomerativeClustering(n_clusters=5, linkage="ward")

df["Cluster"] = model.fit_predict(X)

#Display cluster results

print(df[["Annual Income (k$)", "Spending Score (1-100)", "Cluster"]])

#Visulaize clusters

plt.figure(figsize=(8, 5))

plt.scatter(X.iloc[:, 0], X.iloc[:, 1], c=df["Cluster"], cmap="rainbow")

plt.xlabel("Annual Income")
plt.ylabel("Spending Score")
plt.title("Customer Segmentation")
plt.show()
```

```python
import os
os.environ["OMP_NUM_THREADS"] = "1"

import warnings
warnings.filterwarnings("ignore", category=UserWarning, module="sklearn.cluster._kmeans")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# ==============================
# 1. LOAD + INSPECT
# ==============================
df = pd.read_csv("Mall_Customers.csv")
print(df.shape)
print(df.head())
print(df.describe())
print("Missing values:", df.isnull().sum().sum())

# ==============================
# 2. CHOOSE FEATURES AND SCALE THEM
# ==============================
features = ["Annual Income (k$)", "Spending Score (1-100)"]
X = df[features]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# ==============================
# 3. ELBOW METHOD (+ silhouette score) TO FIND THE BEST K
# ==============================
k_values = range(2, 11)
inertias = []
silhouettes = []

for k in k_values:
    km = KMeans(n_clusters=k, n_init=10, random_state=42)
    km.fit(X_scaled)
    inertias.append(km.inertia_)
    silhouettes.append(silhouette_score(X_scaled, km.labels_))
    print(f"K={k:2d}  inertia={km.inertia_:8.2f}  silhouette={silhouettes[-1]:.3f}")

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].plot(list(k_values), inertias, marker="o")
ax[0].set_xlabel("Number of clusters (K)")
ax[0].set_ylabel("Inertia (WCSS)")
ax[0].set_title("Elbow method (look for the bend)")

ax[1].plot(list(k_values), silhouettes, marker="o", color="green")
ax[1].set_xlabel("Number of clusters (K)")
ax[1].set_ylabel("Silhouette score")
ax[1].set_title("Silhouette score (higher is better)")

plt.tight_layout()
plt.show()

# ==============================
# 4. FIT THE FINAL MODEL
# ==============================
best_k = 5  # <- choose from the elbow / silhouette plots
kmeans = KMeans(n_clusters=best_k, n_init=10, random_state=42)
df["Cluster"] = kmeans.fit_predict(X_scaled)

print("\nCustomers per cluster:\n", df["Cluster"].value_counts().sort_index())
print("Final silhouette score:", round(silhouette_score(X_scaled, df["Cluster"]), 3))

# ==============================
# 5. VISUALIZE THE CLUSTERS
# ==============================
# Centroids were learned in scaled space, so convert them back to real units
centroids = scaler.inverse_transform(kmeans.cluster_centers_)

plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=df,
    x=features[0],
    y=features[1],
    hue="Cluster",
    palette="tab10",
    s=70
)
plt.scatter(
    centroids[:, 0],
    centroids[:, 1],
    marker="X",
    s=250,
    c="black",
    label="Centroids"
)
plt.title(f"K-Means clusters (K={best_k})")
plt.legend()
plt.tight_layout()
plt.show()
```
